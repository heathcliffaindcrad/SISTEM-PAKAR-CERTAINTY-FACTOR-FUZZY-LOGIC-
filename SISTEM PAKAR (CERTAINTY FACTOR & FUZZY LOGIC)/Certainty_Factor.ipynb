CERTAINTY FACTOR

```python
gejala_user = {
    "demam": 0.2,                # Diubah menjadi 0.2
    "batuk": 0.5,
    "sakit_tenggorokan": 0.6,
    # 5 gejala tambahan
    "sakit_kepala": 0.4,
    "nyeri_otot": 0.5,
    "hilang_nafsu_makan": 0.3,
    "mual": 0.2,
    "bersin": 0.4
}

pengetahuan = {
    "flu": {
        "demam": 0.7,
        "batuk": 0.6,
        "sakit_tenggorokan": 0.5,
        "sakit_kepala": 0.5,
        "bersin": 0.6
    },
    "covid": {
        "demam": 0.8,
        "batuk": 0.7,
        "sakit_kepala": 0.6,
        "nyeri_otot": 0.6,
        "hilang_nafsu_makan": 0.7,
        "mual": 0.5
    }
}

def hitung_cf(gejala_user, pengetahuan_pakar):
    cf_total = 0
    first = True
    for gejala, cf_user in gejala_user.items():
        if gejala in pengetahuan_pakar:
            cf_pakar = pengetahuan_pakar[gejala]
            cf = cf_user * cf_pakar  # Perhitungan dasar: CF(pakar) x CF(user)
            if first:
                cf_total = cf
                first = False
            else:
                cf_total = cf_total + cf * (1 - cf_total)
    return cf_total

for penyakit in pengetahuan:
    hasil_cf = hitung_cf(gejala_user, pengetahuan[penyakit])
    print(f"CF diagnosis {penyakit}:  {hasil_cf:.2f}")
```

1. **Data Gejala User**
   - Berisi input gejala dari user beserta nilai kepercayaannya (CF user).
   - Contoh: "demam" = 0.2 artinya user merasa demamnya ringan.

2. **Data Pengetahuan**
   - Berisi nilai CF dari pakar medis (misalnya dokter), menunjukkan seberapa penting gejala tertentu untuk mendiagnosis penyakit tertentu.

3. **Fungsi `hitung_cf()`**
   - Mencocokkan setiap gejala yang dimiliki user dengan yang ada di basis pengetahuan.
   - CF dihitung menggunakan rumus gabungan:  
     `CFtotal = CF1 + CF2 * (1 - CF1)`, untuk menggabungkan dua nilai kepercayaan secara progresif.

4. **Output Diagnosis**
   - Menampilkan hasil akhir CF untuk setiap penyakit.
   - Semakin tinggi nilainya (mendekati 1), semakin yakin sistem bahwa user mengidap penyakit tersebut.


- Walaupun nilai emam diturunkan ke 0.2, diagnosis **flu masih cukup tinggi**, karena ditopang oleh gejala lain seperti `batuk`, `sakit tenggorokan`, dan `bersin`.
- CF untuk covid juga tinggi, karena gejala seperti `nyeri otot`, `mual`, dan `hilang nafsu makan` muncul di input user.

Kesimpulan

> Sistem pakar berbasis CF ini sangat fleksibel karena bisa menyesuaikan diagnosis berdasarkan bobot keyakinan dari user. Saat CF dari satu gejala penting (seperti demam) rendah, sistem tetap bisa memberikan diagnosis akurat jika gejala lain mendukung. CF ini sangat cocok untuk skenario di mana informasi dari user tidak selalu 100% yakin.
